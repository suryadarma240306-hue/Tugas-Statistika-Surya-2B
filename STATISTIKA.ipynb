BERDASARKAN GENRE

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Mengimpor dataset film IMDB dari file CSV
ImdbMovie_df = pd.read_csv('/content/drive/MyDrive/MovieData-2018-2020.csv')

# Menghilangkan data yang tidak memiliki nilai pada kolom 'genre'
# Nilai kosong pada kolom genre dapat menyebabkan proses stratify gagal
ImdbMovie_df = ImdbMovie_df.dropna(subset=['genre'])

# Menyeleksi genre yang memiliki jumlah data minimal 2
# Hal ini dilakukan agar setiap genre masih dapat dibagi saat proses stratified sampling
genre_counts = ImdbMovie_df['genre'].value_counts()
valid_genres = genre_counts[genre_counts >= 2].index
ImdbMovie_df = ImdbMovie_df[ImdbMovie_df['genre'].isin(valid_genres)]
# -------------------------------------

# 2. Membagi data dengan teknik Stratified Sampling
_, stratified_sample = train_test_split(
    ImdbMovie_df,
    test_size=0.3,                     # Menentukan ukuran sampel sebesar 30% dari total data
    stratify=ImdbMovie_df['genre'],    # Menjaga proporsi genre agar tetap seimbang dalam sampel
    random_state=63                    # Mengatur nilai acak berdasarkan NIM agar hasil tetap konsisten
)

# 3. Mengambil 30 baris pertama dari data hasil sampling
sample_30_teratas = stratified_sample.head(30)

# 4. Menampilkan hasil 30 data teratas dari stratified sampling
print("Stratified Sampling (30 Data Teratas):")
display(sample_30_teratas)
```

MENGHITUNG ESTIMASI BUDGET

```python
import scipy.stats as stats
import math

# 1. Mengambil data budget dari hasil 30 sampel pertama
# Periksa kembali agar nama kolom 'budget' sudah sama dengan yang ada pada dataset
data_budget = sample_30_teratas['budget']

# 2. Menghitung nilai statistik utama berdasarkan data sampel
x_bar = data_budget.mean()         # Menentukan nilai rata-rata budget dari sampel
s = data_budget.std(ddof=1)        # Menentukan standar deviasi sampel dengan derajat bebas n-1
n = len(data_budget)               # Menentukan banyaknya data yang digunakan dalam sampel
confidence_level = 0.95            # Menetapkan tingkat kepercayaan sebesar 95%

# 3. Menentukan nilai alpha, derajat bebas, standard error, t kritis, dan margin error
alpha = (1 - confidence_level)
df = n - 1
standard_error = s / math.sqrt(n)
critical_t = stats.t.ppf(1 - alpha/2, df)
margin_of_error = critical_t * standard_error

# 4. Menghitung batas bawah dan batas atas interval kepercayaan
lower_bound = x_bar - margin_of_error
upper_bound = x_bar + margin_of_error

# 5. Menampilkan hasil perhitungan statistik dan confidence interval
print("=== STATISTIK SAMPEL ===")
print("Jumlah Sampel (n)     : ", n)
print("Rata-rata Sampel (x̄)  : {:.2f}".format(x_bar))
print("Standar Deviasi (s)   : {:.2f}".format(s))
print("-" * 30)
print("=== HASIL T-DISTRIBUSI ===")
print("Critical t-value      : {:.2f}".format(critical_t))
print("Margin of error       : {:.2f}".format(margin_of_error))
print("Confidence Interval   : ({:.2f}, {:.2f})".format(lower_bound, upper_bound))
print("The {:.1%} confidence interval for the population mean budget is:".format(confidence_level))
print("Between {:.2f} and {:.2f}".format(lower_bound, upper_bound))
```
